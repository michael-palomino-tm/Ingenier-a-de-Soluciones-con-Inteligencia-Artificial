```python
import logging
import time

# CONFIGURACIÓN PARA GUARDAR EN UN ARCHIVO:
# Limpiamos configuraciones previas de Colab
for handler in logging.root.handlers[:]:
    logging.root.removeHandler(handler)

logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s',
    filename='agente.log',  # <--- Aquí indicamos el nombre del archivo
    filemode='w'           # 'w' sobreescribe el archivo cada vez. Usa 'a' para acumular logs.
)

class Agent:
    def __init__(self):
        self.counter = 0

    def act(self, message):
        start = time.time()
        logging.info(f"Recibido mensaje: {message}")
        self.counter += 1
        response = f"Respuesta #{self.counter} a: {message}"
        duration = time.time() - start
        logging.info(f"Duración de respuesta: {duration:.4f} segundos")
        return response

if __name__ == "__main__":
    agent = Agent()
    # Los 'print' seguirán saliendo en pantalla, pero los logs irán al archivo
    print(agent.act("Hola"))
    print(agent.act("¿Cómo estás?"))
```
